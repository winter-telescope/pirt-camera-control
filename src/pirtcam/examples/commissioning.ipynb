# Notebook for taking data during on-sky commissioning

```python
# Camera Server

from pirtcam.client import CameraClient

# Connect to the camera server
cam = CameraClient("localhost", 5555)
cam.connect()

# Make sure the camera has all corrections turned off
cam.set_correction("GAIN", "OFF")
cam.set_correction("OFFSET", "OFF")   
cam.set_correction("SUB", "OFF")


```

```python
# Set the exposure time

exptime = 1.0  # seconds
cam.set_exposure(exptime)
```

```python
# Set up exposure parameters
n_frames = 10
stack = False  # all frames in a single stack?
```

```python
# Where should the data be saved?
import os
data_dir = os.path.join(os.path.expanduser("~"), "data")

night_dir = os.path.join(data_dir, "commissioning", "20250803")

save_dir = os.path.join(night_dir, "darks")

# SET THE SAVE DIRECTORY
cam.set_save_path(save_dir)
```

```python
filter = "J"
obstype = "DARK"
targname = ""


exptimes = [1.0, 10.0, 30.0]

for exptime in exptimes:
    cam.set_exposure(exptime)
    print(f"Taking {obstype} frame with exposure time {exptime} s")
    
    # capture some dummy frames
    cam.capture_frames(n_frames, stack = stack, headers = {
        "OBSTYPE": obstype,
        "FRMETYP": obstype,
        "FILTER" : filter,
        "TARGNAME": targname,
    }, show_progress = False)
```
