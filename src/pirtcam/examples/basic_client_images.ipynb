## Connect

```python
import os
import time

import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits

from pirtcam.client import CameraClient

cam = CameraClient("localhost", 5555)
cam.connect()

# All corrections OFF - see above.
cam.set_correction("GAIN", "OFF")
cam.set_correction("OFFSET", "OFF")
cam.set_correction("SUB", "OFF")

cam.set_observer_name("Nate Lourie")
cam.set_object_name("DARK")

print(cam.get_status()["data"]["camera_state"])
```

### Set the exposure time

```python
exptime_s = .001

cam.set_exposure(exptime_s)
```

### Set the frame time

```python
cam.set_tec_temperature(15);
```

### Take an exposure

```python
n_frames = 4

header = {
    "FILTER" : "1250",
    "NDFILT" : "2.0",
}

cam.capture_frames(n_frames, 
                   stack = False, 
                   headers = header,
                   show_progress = False)
```
